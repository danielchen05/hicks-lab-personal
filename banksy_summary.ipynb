BANKSY - spatial clustering that unifies cell typing and tissue domain segmentation.  
BANKSY can improve cell-type assignment/ distinguish cell types / identify spatial zones sharing the same microenvironment.  
Looks applicable with Visium HD. Also compatible with all previous analyses using Scanpy.

BANKSY can be used for improving cell type assignment - see slide seq dataset analysis.

The following summarizes information from https://github.com/prabhakarlab/Banksy_py  
R version: https://github.com/prabhakarlab/Banksy  
BANKSY can be run on spatial single-cell expression dataset in anndata format.  
Major steps:  
Preprocess the gene-cell matrix using Scanpy (filter + select most highly variable genes)  
initialize_banksy function to generate spatial graph  
run_banksy_multiparam to perform dimensionality reduction and clustering  

In more details:

Pre-process gene-cell matrix, z-score by gene (standardize). Processed by scanpy.

construct spatial graph (defines spatial neighbor relationship). output sparse adjacency matrix. PARAMETERS to be modified: k-nearest neighbors (to connect graph to), distributions of weights of a cell's neighbor expression (gaussian, uniform = weigh all neighbor equally according to distance). Spatial graph can be generated, which can show # neighbors per cell.

generate neighbor expression matrix to average over spatial neighbors. Computation: sparse matrix multiplication of adjacency matrix with gene-cell matrix.

scale original expression matrix, concatenate to get neighbor-augmented expression matrix. (?)

Other processing step relevant to single cell RNA seq - PCA, graph-based clustering (Leiden), can REFINE noisy clusters using functions.

**Special Adaptations for Visium HD**

BANKSY at lambda = 0 for non-spatial clustering. 0.2 for cell-typing. 
For Visium HD, set lambda = 0.2 for cell typing, 0.8 for domain segmentation.
