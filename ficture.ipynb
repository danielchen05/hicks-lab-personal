This notebook goes over the tool FICTURE, which is used to compare results with bin2cell.  
Paper link: https://www.nature.com/articles/s41592-024-02415-2  
Github link: https://github.com/seqscope/ficture  
Documentation link: https://seqscope.github.io/ficture/

Segmentation-free analysis of submicro-resolution analysis of ST data.  
- Identify cell types from high-resolution data without compromising its original resolution
- does not require external cell segmentation
- Can switch between 2 modes (all in one = unsupervised clustering + pixel-level decoding, projection = pixel-level decoding from externally labeled cell types)

Input requirement: .tsv file format (only needs spatial coordinates and expression values), has a workflow developed for visium HD data already at this link:  
https://seqscope.github.io/ficture/format_input/visiumHD/

Using FICTURE: a pipeline is already developed and can run all steps together https://seqscope.github.io/ficture/quickstart/

A more detailed process and algorithm:  
https://seqscope.github.io/ficture/localrun/

Input data: 
- transcript file: contain molecular or pixel level information, requires X,Y coordinates, gene, and count (# of transcripts observed)
- gene list file: contains unique names of genes used in analysis
- bounding box/area for spatial coordinates in um.

Setup for analysis:
 - mu_scale: ratio between um and unit used in transcript coordinates
 - key: column name that corresponds to gene counts
 - major axis: the transcript file needs to be sorted - the axis the file sorted by

Pre-processing:
- create pixel minibatches for pixel level analysis (split data into subsets of pixels)
- prepare training hexagons (ensures spatial coherence - each hexagon defines a neighborhood, similar expression patterns should be observed in a hexagon)

Model training:
- uses LDA (Latent Dirichlet Allocation), topic modeling for gene expression, unsupervised

LDA is trained to learn: Per-factor gene distributions (which genes are common in each factor), Per-pixel topic mixtures (how much of each factor is expressed at each location, factor proportions).

Pixel-level decoding:
- to infer factors (expression patterns) for each pixel
- Step 1: perform anchor-level projection based on fitted model (factor profiles assigned to training/anchor hexagons)
- Step 2: perform pixel-level decoding based on anchor-level projection (for each pixel, factor representation estimated by interpolating from nearby anchors, and use gene counts to adjust projection)

Post-Processing:
- identify differentially expressed genes
- generate heatmaps

Main results:
- spatial factor map: where each inferred factor (spatial gene expression pattern) is active, how strongly each pixel expresses each factor.
